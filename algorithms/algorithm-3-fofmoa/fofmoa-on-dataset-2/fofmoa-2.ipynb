```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split, KFold
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (accuracy_score, precision_score, recall_score, f1_score,
                             balanced_accuracy_score, confusion_matrix,
                             matthews_corrcoef, cohen_kappa_score, classification_report)
import time
import os

# Set the number of K folds as a global variable
K_FOLDS = 5

# Read the dataset from CSV file
df = pd.read_csv('../../../datasets/dataset-2.csv')

# Explore dataset
print("Dataset shape:", df.shape)
print("\nDataset columns:", df.columns.tolist())
print("\nClass distribution:")
print(df['label'].value_counts())

# Define features and target
X = df.drop(columns=['label']).values
y = df['label'].values

# Create directory for outputs
os.makedirs("confusion_matrices", exist_ok=True)

# Store timing results
timing_results = []

# Define classifiers
classifiers = {
    'DecisionTree': DecisionTreeClassifier(random_state=42),
    'KNN': KNeighborsClassifier(n_neighbors=5),
    'LogisticRegression': LogisticRegression(random_state=42, max_iter=1000, multi_class='multinomial'),
    'RandomForest': RandomForestClassifier(random_state=42, n_estimators=100)
}

# Store results
results = []

# Create K-fold cross-validation
kf = KFold(n_splits=K_FOLDS, shuffle=True, random_state=42)

# Helper function for confusion matrix metrics
def confusion_matrix_metrics(cm, classes):
    metrics = {}
    for idx, class_label in enumerate(classes):
        TP = cm[idx, idx]  # True Positives for this class
        FP = cm[:, idx].sum() - TP  # False Positives for this class
        FN = cm[idx, :].sum() - TP  # False Negatives for this class
        TN = cm.sum() - (TP + FP + FN)  # True Negatives for this class
        
        metrics[class_label] = {
            'TPR': TP / (TP + FN + 1e-10) if (TP + FN) > 0 else 0,  # Sensitivity/Recall
            'TNR': TN / (TN + FP + 1e-10) if (TN + FP) > 0 else 0,  # Specificity
            'FPR': FP / (FP + TN + 1e-10) if (FP + TN) > 0 else 0,  # False Positive Rate
            'FNR': FN / (FN + TP + 1e-10) if (FN + TP) > 0 else 0   # False Negative Rate
        }
    return metrics

# Iterate over classifiers
for clf_name, clf in classifiers.items():
    print(f"\nTraining {clf_name}...")
    fold_idx = 1
    
    # Store metrics for each class across folds
    class_metrics_by_fold = {}
    
    for train_index, test_index in kf.split(X):
        print(f"  Processing fold {fold_idx}/{K_FOLDS}")
        
        # Split the data
        X_train, X_test = X[train_index], X[test_index]
        y_train, y_test = y[train_index], y[test_index]
        
        # Record start time
        start_train_time = time.time()
        clf.fit(X_train, y_train)
        train_time = time.time() - start_train_time
        
        start_test_time = time.time()
        y_pred = clf.predict(X_test)
        test_time = time.time() - start_test_time
        
        timing_results.append({
            'Classifier': clf_name,
            'Fold': fold_idx,
            'Training Time (s)': train_time,
            'Testing Time (s)': test_time,
            'Total Time (s)': train_time + test_time
        })
        
        # Get the actual classes that are present in the test data
        actual_test_classes = np.unique(np.concatenate([y_test, y_pred]))
        
        # Compute metrics
        cm = confusion_matrix(y_test, y_pred, labels=actual_test_classes)
        cm_metrics = confusion_matrix_metrics(cm, actual_test_classes)
        
        # For multi-class, calculate overall metrics
        overall_metrics = {
            'Classifier': clf_name,
            'Fold': fold_idx,
            'Class': 'Overall',
            'Accuracy': accuracy_score(y_test, y_pred),
            'Precision': precision_score(y_test, y_pred, average='weighted', zero_division=0),
            'Recall': recall_score(y_test, y_pred, average='weighted', zero_division=0),
            'F1 Score': f1_score(y_test, y_pred, average='weighted', zero_division=0),
            'Balanced Accuracy': balanced_accuracy_score(y_test, y_pred),
            'Matthews Correlation': matthews_corrcoef(y_test, y_pred),
            'Cohen Kappa': cohen_kappa_score(y_test, y_pred),
            'Training Time (s)': train_time,
            'Testing Time (s)': test_time
        }
        results.append(overall_metrics)
        
        # Calculate class-specific metrics
        for class_label in actual_test_classes:
            # Create binary indicators for the current class (one-vs-rest approach)
            y_test_binary = np.where(y_test == class_label, 1, 0)
            y_pred_binary = np.where(y_pred == class_label, 1, 0)
            
            class_specific_metrics = {
                'Classifier': clf_name,
                'Fold': fold_idx,
                'Class': f'Class {class_label}',
                'Accuracy': accuracy_score(y_test_binary, y_pred_binary),
                'Precision': precision_score(y_test_binary, y_pred_binary, zero_division=0),
                'Recall': recall_score(y_test_binary, y_pred_binary, zero_division=0),
                'F1 Score': f1_score(y_test_binary, y_pred_binary, zero_division=0),
                'True Positive Rate (TPR)': cm_metrics[class_label]['TPR'],
                'True Negative Rate (TNR)': cm_metrics[class_label]['TNR'],
                'False Positive Rate (FPR)': cm_metrics[class_label]['FPR'],
                'False Negative Rate (FNR)': cm_metrics[class_label]['FNR'],
                'Training Time (s)': train_time,
                'Testing Time (s)': test_time
            }
            results.append(class_specific_metrics)
            
        # Plot and save confusion matrix
        plt.figure(figsize=(10, 8))
        sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', 
                   xticklabels=[f'Class {c}' for c in actual_test_classes], 
                   yticklabels=[f'Class {c}' for c in actual_test_classes])
        plt.title(f"{clf_name} - Fold {fold_idx} Confusion Matrix")
        plt.xlabel("Predicted")
        plt.ylabel("True")
        plt.tight_layout()
        plt.savefig(f"confusion_matrices/{clf_name}_fold_{fold_idx}.png")
        plt.close()
        
        # Print classification report for this fold
        # Only use the classes that actually appear in the test data
        target_names = [f'Class {c}' for c in actual_test_classes]
        print(f"\nClassification Report for {clf_name} - Fold {fold_idx}:")
        print(classification_report(y_test, y_pred, labels=actual_test_classes, target_names=target_names))
        
        fold_idx += 1

# Create DataFrames for results
timing_df = pd.DataFrame(timing_results)
results_df = pd.DataFrame(results)

# Save timing results
timing_df.to_csv("time.csv", index=False)
print("\nTiming Results:")
print(timing_df.groupby('Classifier')[['Training Time (s)', 'Testing Time (s)', 'Total Time (s)']].mean())

# Save metrics results
results_df.to_csv("metrics.csv", index=False)
print("\nClassification Metrics Summary:")
numeric_cols = ['Accuracy', 'Precision', 'Recall', 'F1 Score', 'Balanced Accuracy', 
                'Matthews Correlation', 'Cohen Kappa', 'Training Time (s)', 'Testing Time (s)']
overall_results = results_df[results_df['Class'] == 'Overall']
summary_metrics = {}

for clf in overall_results['Classifier'].unique():
    clf_data = overall_results[overall_results['Classifier'] == clf]
    summary_metrics[clf] = {col: clf_data[col].mean() for col in numeric_cols if col in clf_data.columns}

summary_df = pd.DataFrame(summary_metrics).T
print(summary_df)

# Create visualizations for overall results
plt.figure(figsize=(14, 8))

# Accuracy comparison
plt.subplot(2, 2, 1)
sns.barplot(x='Classifier', y='Accuracy', data=results_df[results_df['Class'] == 'Overall'])
plt.title('Accuracy by Classifier')
plt.ylim(0, 1)

# F1 Score comparison
plt.subplot(2, 2, 2)
sns.barplot(x='Classifier', y='F1 Score', data=results_df[results_df['Class'] == 'Overall'])
plt.title('F1 Score by Classifier')
plt.ylim(0, 1)

# Balanced Accuracy comparison
plt.subplot(2, 2, 3)
sns.barplot(x='Classifier', y='Balanced Accuracy', data=results_df[results_df['Class'] == 'Overall'])
plt.title('Balanced Accuracy by Classifier')
plt.ylim(0, 1)

# Training Time comparison
plt.subplot(2, 2, 4)
sns.barplot(x='Classifier', y='Training Time (s)', data=timing_df)
plt.title('Training Time by Classifier')

plt.tight_layout()
plt.savefig("performance_comparison.png")
plt.close()

# Class-specific performance for best classifier
f1_means = {}
for clf in results_df['Classifier'].unique():
    clf_data = results_df[(results_df['Classifier'] == clf) & (results_df['Class'] == 'Overall')]
    if not clf_data.empty:
        f1_means[clf] = clf_data['F1 Score'].mean()

if f1_means:
    best_clf = max(f1_means, key=f1_means.get)
    class_metrics = results_df[(results_df['Classifier'] == best_clf) & 
                               (results_df['Class'] != 'Overall')]

    plt.figure(figsize=(12, 8))
    sns.barplot(x='Class', y='F1 Score', data=class_metrics)
    plt.title(f'F1 Score by Class for {best_clf}')
    plt.ylim(0, 1)
    plt.tight_layout()
    plt.savefig(f"{best_clf}_class_performance.png")

    print(f"\nBest performing classifier: {best_clf}")
else:
    print("\nCould not determine best classifier due to missing data")

print("Done! All results saved to files.")
```
